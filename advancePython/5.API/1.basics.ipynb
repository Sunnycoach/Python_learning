```python
import requests
import json
```

```python
url = "https://jsonplaceholder.typicode.com/posts/20"

response = requests.get(url)
print(response.json())
```

```python
print(json.dumps(response.json(), indent=1))
```

```python
url1 = "https://jsonplaceholder.typicode.com/posts"

data = {
    "title": "foo",
    "body": "bar",
    "userId": 1
}
r1 = requests.post(url1,json=data)
```

```python
print(json.dumps(r1.json(), indent=1))
```

```python
form_data = {
    "title": "xyz",
    "body": "rab",
    "userId": 10
}

response2 = requests.post(url1, data=form_data)
```

```python
print(json.dumps(response2.json(), indent=1))
print(f"status code: {response2.status_code}")
```

```python
#QUERY PARAMATERS

url2 = "https://jsonplaceholder.typicode.com/posts"

params = {
    "userId": 1,
    "id": 1
}

response3 = requests.get(url2, params=params)
print(json.dumps(response3.json(), indent=1))
print(f"status code: {response3.status_code}")
```
