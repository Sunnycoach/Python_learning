```python
s = "A man, a plan, a canal: Panama"
left = 0
right = len(s)-1

while left < right:
    while left < right and not s[left].isalnum():
        left += 1
    
    while left < right and not s[right].isalnum():
        right -= 1

    if s[left].lower() != s[right].lower():
        print("no")
        break
    left += 1
    right -= 1
else:
    print("yes")
```

❓ Problem:

Compress string

```python
chars = ["a","a","b","b","c","c","c","k","d","d",'b','b']

write = 0
read = 0

while read < len(chars):
    char = chars[read]
    count = 0

    while read < len(chars) and chars[read] == char:
        read +=1
        count +=1

    chars[write] = char
    write += 1

    if count > 1:
        for i in str(count):
            chars[write] = i
            write += 1
            
print(chars[:write])
```

```python
s = "abcdabcbb"

char_set = set()
left = 0
max_len = 0

for r in range(len(s)):
    while s[r] in char_set:
        char_set.remove(s[left])
        left+=1
    char_set.add(s[r])
    max_len = max(max_len, r - left + 1)
print(max_len)
```
