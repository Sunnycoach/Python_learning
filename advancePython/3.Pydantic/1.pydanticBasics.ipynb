```python
from pydantic import BaseModel, StrictInt
```

```python
class employeeData(BaseModel):
    name:str
    age:int
    dept:str
```

```python
obj=employeeData(name="Sunny",age=23,dept="MSU")
obj
```

```python
obj2=employeeData(**{"name": "Sunny", "age": 23, "dept": "101"})
obj2
```

```python
obj2=employeeData(**{"name": "Sunny", "age": "23", "dept": "101"})
obj2
```

```python
obj4=employeeData(**{"name": "Sunny", "age":"23", "dept": "101"})
obj4
```

```python
class employeeData(BaseModel):
    name:str
    age:StrictInt
    dept:str
```

```python
obj5=employeeData(**{"name": "Sunny", "age":23, "dept": "101"})
obj5
```
